```python
# 1.
# START

# INPUT a string s

# SPLIT the string into individual words

# CREATE an empty dictionary

# FOR each word in the list:
#     IF the word already exists in the dictionary:
#         INCREASE its count by 1
#     ELSE:
#         SET its count to 1

# RETURN the dictionary

# END

def wrd_freq(s):
    s2 = s.split()
    wrd_d = {}

    for i in s2:
        if i in wrd_d:
            wrd_d[i]+=1
        else:
            wrd_d[i]=1

    return wrd_d
```

```python
# 2.
# START

# INPUT a string w

# CREATE an empty dictionary

# FOR each character in the string:
#     IF the character already exists in the dictionary:
#         INCREASE its count by 1
#     ELSE:
#         SET its count to 1

# RETURN the dictionary

# END


def freq_ch(w):

    ch_d = {}
    for i in w:
        if i in ch_d:
            ch_d[i]+=1
        else:
            ch_d[i]=1

    return ch_d    
```

```python
# 3.
# START

# INPUT a string w

# CREATE an empty dictionary
# SET x to a blank space

# FOR each character in the string:
#     IF the character exists in the dictionary:
#         INCREASE its count by 1
#     ELSE:
#         SET its count to 1

# FOR each character and its count in the dictionary:
#     IF the count equals 1:
#         SET x to that character
#         STOP the loop

# RETURN x

# END


def non_rep(w):
    n_dic = {}
  #  lst = []
    x = " "
    for i in w:
        if i in n_dic:
            n_dic[i]+=1
        else:
            n_dic[i]=1

    for k,v in n_dic.items():
        if v == 1:
            x = k        
            break

    return x
```

```python
# 4.
# START

# INPUT two strings s1 and s2

# CONVERT both strings to lowercase

# IF the lengths of s1 and s2 are not equal:
#     RETURN False

# FOR each character in s1:
#     SET c1 = 0
#     SET c2 = 0

#     COUNT occurrences of the character in s1
#     STORE the count in c1

#     COUNT occurrences of the character in s2
#     STORE the count in c2

#     IF c1 is not equal to c2:
#         RETURN False


def st_agrm(s1,s2):
    s1 = s1.lower()
    s2 = s2.lower()


    if len(s1)!=len(s2):
        return False

    for ch in s1:
        c1 = 0
        c2 = 0

        for i in s1:
            if i == ch:
                c1+=1

        for i in s2:
            if i == ch:
                c2+=1

        if c1!=c2:
            return False                
    return True
```

```python
s_anagram = st_agrm("silent","listen")
print(s_anagram)
```

```python
# 5. 

# START

# INPUT a string s

# CREATE an empty dictionary
# SET x to a blank space

# FOR each character in the string:
#     IF the character exists in the dictionary:
#         INCREASE its count by 1
#     ELSE:
#         SET its count to 1

# FOR each character and its count in the dictionary:
#     IF the count is greater than 1:
#         SET x to that character
#         STOP the loop

# RETURN x

# END


def first_repeat(s):
    char_f = {}
    x = " "

    for i in s:
        if i in char_f:
            char_f[i]+=1
        else:
            char_f[i]=1

    for k,v in char_f.items():
        if v > 1:
            x = k
            break

    return k   
```

```python
#6. 

# START

# Initialize nums with the given list
# Create an empty result list

# FOR each element i in nums:
#     IF i is not equal to 0:
#         Append i to result

# Calculate zeros = length of nums - length of result

# REPEAT zeros times:
#     Append 0 to result

# PRINT result

# END



def mv_zeros(nums):
    lst = []

    for i in nums:
        if i!=0:
            lst.append(i)

    z = len(nums) - len(lst)

    for i in range(1,z+1):
        lst.append(0)            

    return lst  
```

```python
# 9.
# START

# INPUT a list nums

# FIND the length of the list and store it in n

# CALCULATE the expected sum using n * (n + 1) / 2
# STORE it in e_sum

# CALCULATE the actual sum of the list
# STORE it in a_sum

# SUBTRACT the expected sum from the actual sum
# STORE the result in m_num

# RETURN m_num

# END




def miss_num(nums):
    n = len(nums)
    e_sum = n * (n+1) // 2
    a_sum = sum(nums)
        
    m_num = a_sum - e_sum
        
    return m_num    
```

```python
miss_num([1, 2, 4, 5, 6])
```

```python
# 10.

# START

# INPUT a list nums and a target value tar

# CREATE an empty list lst

# FOR i from 0 to the length of nums - 1:
#     FOR j from i + 1 to the length of nums - 1:

#         IF nums[i] + nums[j] equals tar:
#             ADD nums[i] to lst
#             ADD nums[j] to lst

# RETURN lst

# END

def sum_pair(nums,tar):
    lst = []
    for i in range(len(nums)):
        for j in range(i+1,len(nums)):
            if nums[i] + nums[j] == tar:
                # print(nums[i],nums[j])
                lst.append(nums[i])
                lst.append(nums[j])
    return lst            
```

```python
l = [2, 7, 11, 15]
k = 9
sum_pair(l,k)
```

```python
#13. 

# START

# Initialize nums with the given list
# Initialize K with the number of positions to rotate

# Set n = length of nums
# Set K = K modulo n

# Create an empty result list

# FOR i from n - K to n - 1:
#     Append nums[i] to result

# FOR i from 0 to n - K - 1:
#     Append nums[i] to result

# RETURN result

# END


def rotate_right(nums, k):
    n = len(nums)
    k = k % n

    result = []

    for i in range(n - k, n):
        result.append(nums[i])

    for i in range(0, n - k):
        result.append(nums[i])

    return result
```

```python
nums = [1, 2, 3, 4, 5]
k = 2

print(rotate_right(nums, k))
```

```python
# 14.


# START

# Initialize nums with the given list

# Set max_sum = nums[0] + nums[1]

# FOR i from 1 to length of nums - 2:

#     Set curr_sum = nums[i] + nums[i + 1]

#     IF curr_sum > max_sum:
#         Set max_sum = curr_sum

# RETURN max_sum

# END


def con_sum(lst):
    max_sum = lst[0] + lst[1]

    for i in range(1, len(lst)-1):
        curr_sum = lst[i] + lst[i+1]

        if curr_sum > max_sum:
            max_sum = curr_sum

    return max_sum      
```

```python
l = [4, 7, 2, 9, 5]
print(con_sum(l))
```

```python
# 15. 

# START

# Initialize nums with the given list
# Initialize target with the required sum
# Create an empty list called pairs

# FOR i from 0 to length of nums - 1:

#     FOR j from i + 1 to length of nums - 1:

#         IF nums[i] + nums[j] equals target:

#             Set a = nums[i]
#             Set b = nums[j]

#             IF a is greater than b:
#                 Swap a and b

#             Create pair = [a, b]

#             IF pair is not already in pairs:
#                 Append pair to pairs

# Sort pairs in ascending order

# RETURN pairs

# END





def sum_target(nums, target):
    pairs = []

    for i in range(len(nums)):
        for j in range(i + 1, len(nums)):
            if nums[i] + nums[j] == target:
                a = nums[i]
                b = nums[j]

                if a > b:
                    a, b = b, a

                pair = [a, b]

                if pair not in pairs:
                    pairs.append(pair)

    pairs.sort()
    return pairs
```
