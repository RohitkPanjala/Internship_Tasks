## Q1. Find the Shortest Word in a Sentence
Write a Python program to find the shortest word in a given sentence.

Example input:
Python makes coding easy

Expected output:
easy

```python
s = "Python makes coding easy"  # The input for the code is string and output is also a string

s2 = s.split(" ")       # I split the string with spaces using 'split' function and created a list s2. So I can access words in sentence indivdually

shortest_wrd = s2[0]      # assumed first index value in list as shortest word

for i in s2:             # using for loop, I itterated every string value in list for each word. Then I compared every word in with shortest word.
    if len(i) < len(shortest_wrd):       # if any word satisfies condition, shortest_wrd is updated until the end
        shortest_wrd = i

        
print(f"shortest word: {shortest_wrd}")         # finally after satisfying every word in list, the final updated shortest_wrd is printed
```

Write a Python program to find the largest number in a list without using the built-in max()
function.

Example input:

```python
lst = [12, 45, 7, 89, 23] # the input is a list and output will be an 'int' value
large_num = lst[0] # firstly we should assume the first index value as largest value

for i in lst: # using 'for' loop we have iterate through all list values
    if i > large_num: # any number which is greater than the number in large_num will be updated until the end of list
        large_num = i       

print(f"The largest number in list is: {large_num}") # finally the last updated value which satisfies our usecase will be printed here
```

## Q6. Count Vowels in a String

Write a Python program to count the total number of vowels in a string. Consider both
uppercase and lowercase vowels.

Example input:
Education

Expected output:
5

```python
s = "Education" # the input is string and output is an integer. Here we need to count vowels
vwls = "aeiou" # as vowels contain letters "aeiou", i created a string withn vowels to compare every word in string

c = 0 # created a variable c as int datatype initially assuming as 0, so as we can increase the count as we find a vowel

for i in s.lower(): # using for loop we have iterate every value of s and with lower so as to avoid case sensitivity
    if i in vwls: # if i value found in vowels(vwls) the we have to increment count with 1
        c+=1


print(f"Number of vowels: {c}") # after iterating through every word finally print the count c for vowels
```

## Q8. Find the Most Frequent Character

Write a Python program to find the character that appears most frequently in a string. If multiple
characters have the same highest frequency, return the first one encountered. Ignore spaces.

Example input:
banana

Expected output:
a

```python
s = "banana"     # here the input is string and output is also string

frq_chr = {}          # we have to check the most frequent character, so in order to keep count we have to use dictionary which is easy to keep track

for i in s:            # iterate every char from string 's' using for loop
    if i in frq_chr:        # if the char found and already in dictionary we have to increament its value with 1
        frq_chr[i]+=1     # since we using dictionary key-value method and here key is char and value is frequecy count of word in string
    else:
        frq_chr[i] = 1      # if char is not in dictionary then make its value as 1

max_ch = s[0]             # so to find the most frequent char, we create a variable called max_char and assume first char as max from given string
max_count = 0             # this is the count of chars apperance in string

for k,v in frq_chr.items():   # iterating every dictionary key-values 
    if v > max_count:         # if the v which is value is greater than max_count along with max_ch which is character, everytime the value gets updated until the loop finsihes iteration
        max_count = v  
        max_ch = k
        

print(max_ch)               # finally print the final max_ch in output        
```

## Q15. Create a Dictionary of Numbers and Their Even/Odd Status

Write a Python program to create a dictionary containing each number as a key and "Even" or "Odd" as its value.

Example input:
[1, 2, 3, 4, 5]

Expected output:
{1: 'Odd', 2: 'Even', 3: 'Odd', 4: 'Even', 5: 'Odd'}

```python
lst = [1, 2, 3, 4, 5]    # the input is a list and output is a dictionary

n_dic = {} # created an empty dictionary to print the output

for i in lst: # using for loop, iterated each int value from list to check if number is even or odd 
    if i%2==0: # if the value is even, the value will be as "Even" or else "Odd"
        n_dic[i]="Even"
    else:
        n_dic[i]="Odd"

print(n_dic) # printed the final output dictionary
```

## Q16. Find the Most Frequent Word in a Sentence

uppercase and lowercase versions of a word as the same word. If multiple words share the

Expected output:
